```python
import numpy as np
import pandas as pd

print("Numpy version :  ",np.__version__)
print("Pandas version : ",np.__version__)
```

```python
marks  = np.array([20,50,47])

print("Marks: ",marks)
print("type: ",type(marks))
```

```python
print("Number of dimensions: ",marks.ndim)
print("Shape: ",marks.shape)
print("Number of elements: ",marks.size)
```

```python
marks  = np.array([20,50,47])
print("Add 5 : ",marks+5)
print("Multiply by 2: ",marks*2)
print("Divide by 2: ",marks/2)
print("Square : ",marks**2)
```

```python
print("Sum: ",marks.sum())
print("Max: ",marks.max())
print("Min: ",marks.min())
print("Mean: ",marks.mean())
```

## Access through Index

```python
print("First marks: ",marks[0])
print("Second marks: ",marks[-2])
print("Last marks: ",marks[-1])
```

### Slicing

```python
print("Index is 1 to 3: ",marks[1:3])
print("First three: ",marks[:3])
print("From Index 2: ",marks[2:])
print("Reverse: ",marks[::-1])
```

### Two-D Arrays

```python
st_marks = np.array([
    [10,20,30],
    [40,50,50],
    [60,70,80]
])
print(st_marks)
print("Shape : ",st_marks.shape)
```

```python
print("Mean of all value: ",st_marks.mean())
print("Column wise mean: ",st_marks.mean(axis=0)) # axis=0 mean calculated downwards
print("Row wise mean: ",st_marks.mean(axis=1)) # axis=1 mean calculates row-wise
```

```python
data={
    "Name":["Ram","Raj","James","Louis"],
    "Branch":["CSE","AIML","CSE","ISE"],
    "Marks":[90,89,78,49]
}
df = pd.DataFrame(data)
df
df.info()
```

```python
df.describe()
```

```python
print("Marks Column ")
print(df["Marks"])
print("\n Branch and Marks")
print(df[["Branch","Marks"]])
```

Row Indexing

```python
print("First row-")
print(df.iloc[0])

print("\n Second row-")
print(df.iloc[1])

print("\n Third row-")
print(df.iloc[2])

```

```python
print("First two rows and first two columns: ")
print(df.iloc[0:2,0:2])
```

Data Frame Slicing

```python
print("First three rows: ")
display(df.iloc[0:3])
print("Rows 1 to 3 , Name and Marks: ")
display(df.iloc[1:4,0:3])
```

Filtering the Data

```python
high_scores = df[df["Marks"]>85]
high_scores
```

```python
print("Students from CSE: ")
display(df[df["Branch"] == "CSE"])

print("Students scoring 85 or above:")
display(df[df["Marks"]>= 85])
```
